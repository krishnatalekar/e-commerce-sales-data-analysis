```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('Sample - Superstore.csv', encoding = 'latin-1')
```

```python
data.head()
```

```python
data.describe()
```

```python
data.info()
```

## Converting date columns

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])
data['Ship Date'] = pd.to_datetime(data['Ship Date'])
```

```python
data.info()
```

```python
data.head()
```

```python
data['Order Month'] = data['Order Date'].dt.month
data['Order Year'] = data['Order Date'].dt.year
data['Order Day of Week'] = data['Order Date'].dt.dayofweek
```

```python
data.head()
```

# 1. Monthly sales analysis

```python
sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index()
```

```python
sales_by_month
```

```python
plt.figure(figsize=(10,6))

sns.lineplot(
    data=sales_by_month,
    x='Order Month',
    y='Sales',
    marker=0,
    linewidth=2.5
)

plt.title('Monthly Sales Analysis', fontsize=18)
plt.xlabel('Order Month', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.4)

plt.show()
```

```python
data.head()
```

# 2. Sales by Category

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()
```

```python
sales_by_category
```

```python
labels = sales_by_category['Category']
sizes = sales_by_category['Sales']

colors = ['#66c2cc', '#ffd166', '#ff9f6e']

plt.figure(figsize=(8,8))

plt.pie(
    sizes,
    labels = labels,
    colors = colors,
    autopct = '%1.1f%%',
    startangle = 90,
    pctdistance = 0.75,
    wedgeprops = dict(width = 0.4, edgecolor = 'white')
)

plt.title('Sales Analysis by Category', fontsize = 20)

plt.tight_layout()
plt.show()
```

# 3. Sales Analysis by Sub-Category

```python
data.head()
```

```python
sales_by_sub_category = data.groupby('Sub-Category')['Sales'].sum().reset_index()
```

```python
sales_by_sub_category
```

```python
plt.figure(figsize=(14,6))

sns.barplot(
    data = sales_by_sub_category,
    x = 'Sub-Category',
    y = 'Sales',
    palette = 'viridis'
)

plt.title('Sales by Sub-Category', fontsize = 18, fontweight = 'bold')
plt.xlabel('Sub-Category')
plt.ylabel('Sales')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

# 4. Monthly Profit Analysis

```python
data.head()
```

```python
profit_by_month = data.groupby('Order Month')['Profit'].sum().reset_index()
```

```python
profit_by_month
```

```python
sns.set_style('whitegrid')

plt.figure(figsize=(12,6))

sns.lineplot(
    data = profit_by_month,
    x = 'Order Month',
    y = 'Profit',
    marker = 'o',
    linewidth = 3,
    color = 'green'
)

plt.title('Monthly Profit Analysis', fontsize = 20, fontweight = 'bold')
plt.xlabel('Order Month', fontsize = 12)
plt.ylabel('Profit', fontsize = 12)

plt.tight_layout()
plt.show()
```

# 5. Profit by Category

```python
profit_by_category = data.groupby('Category')['Profit'].sum().reset_index()
```

```python
profit_by_category
```

```python
colors = sns.color_palette('pastel')

plt.figure(figsize=(8,8))

plt.pie(
    profit_by_category['Profit'],
    labels = profit_by_category['Category'],
    autopct = '%1.1f%%',
    startangle = 90,
    colors = colors,
    pctdistance = 0.75,
    wedgeprops = dict(width = 0.45, edgecolor = 'white')
)

plt.title(
    'Profit Analysis by Category',
    fontsize = 18,
    fontweight = 'bold'
)

plt.tight_layout()
plt.show()
```

# 6. Profit by Sub-Category

```python
data.head()
```

```python
profit_by_sub_category = data.groupby('Sub-Category')['Profit'].sum().reset_index()
```

```python
profit_by_sub_category
```

```python
plt.figure(figsize=(14,6))

sns.barplot(
    data = profit_by_sub_category,
    x = 'Sub-Category',
    y = 'Profit'
)

plt.title('Profit Analysis by Sub-Category', fontsize = 18, fontweight = 'bold')
plt.xlabel('Sub-Category', fontsize = 12)
plt.ylabel('Profit', fontsize = 12)

plt.xticks(rotation = 45)

plt.tight_layout()
plt.show()
```

# 7. Sales and Profit - Customer Segment

```python
data.head()
```

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
```

```python
sales_profit_by_segment
```

```python
plt.figure(figsize=(10,6))

plot_data = sales_profit_by_segment.melt(
    id_vars = 'Segment',
    value_vars = ['Sales', 'Profit'],
    var_name = 'Metric',
    value_name = 'Amount'
)

sns.barplot(
    data = plot_data,
    x = 'Segment',
    y = 'Amount',
    hue = 'Metric',
    palette = 'pastel'
)

plt.title('Sales and Profit Analysis by Customer Segement', fontsize = 18, fontweight = 'bold')
plt.xlabel('Segment', fontsize = 12)
plt.ylabel('Amount', fontsize = 12)
plt.legend(title = '')
plt.grid(axis = 'y', linestyle = '--', alpha = 0.3)

plt.show()
```

# 8. Sales to Profit Ratio

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment', 'Sales_to_Profit_Ratio']])
```

# Conclusion

This project analyzed an e-commerce sales dataset to uncover key business insights related to sales performance, profitability, product categories, and customer segments.

**Key Findings**

* Monthly Sales Analysis: November recorded the highest sales, while
February had the lowest sales, highlighting seasonal demand patterns.
* Category Performance: The Technology category generated the highest share of sales (36.4%), whereas Office Supplies contributed the lowest (31.3%).
* Sub-Category Analysis: Phones emerged as the highest-selling product sub-category.
* Monthly Profit Analysis: December generated the highest overall profit, indicating strong year-end business performance.
* Profit by Category and Sub-Category: The Technology category contributed the highest profit (50.8%), while Copiers proved to be the most profitable product sub-category.
* Customer Segment Analysis: The Consumer segment generated both the highest sales and the highest profit among all customer segments.
* Sales-to-Profit Analysis: The Consumer segment also achieved the strongest sales-to-profit ratio, making it the most valuable customer segment.

**Business Value**

The insights generated from this analysis can help business stakeholders make more informed, data-driven decisions by:

* Optimizing inventory levels based on seasonal sales trends.
* Increasing marketing efforts during high-performing months while developing strategies to improve sales during low-performing periods.
* Prioritizing investment in high-performing categories such as Technology and profitable products like Copiers and Phones.
* Reviewing the performance of lower-performing categories, such as Office Supplies, to identify opportunities for pricing, promotions, or product assortment improvements.
* Designing targeted marketing campaigns for the Consumer segment, which delivers the highest revenue and profitability.
* Improving overall profitability by allocating resources toward products and customer segments that generate the greatest return on investment.
